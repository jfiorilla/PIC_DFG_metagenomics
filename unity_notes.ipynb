## Beginner tips
- Only code run in ipnyb is r; all the bash script is just included for posterity and is actually run in terminal ssh unity 
- Need to create [conda environment](https://unityhpc.org/documentation/software/conda/) with packages - unity uses miniforge rather than miniconda


## Conda code bits
- Load conda: `module load conda/latest`
- Create environment: `conda create --name XXX`
- Activate environment: `conda activate XXX`

## Git

- Clone repo in unity: `git clone <url>`
    - Will need to [create personal token in lieu of password](https://www.geeksforgeeks.org/git/how-to-fix-support-for-password-authentication-was-removed/)
- Update unity directory to reflect changes in repo: `git pull`
